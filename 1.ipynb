```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# nltk.download('punkt_tab')
# nltk.download('stopwords')

text = "Hey! I am learning a new thing today and I am excited"
print(text)

tokens = word_tokenize(text)
print(tokens)

stopwords_list = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word.lower() not in stopwords_list]
print(filtered_tokens)
```

```python
from nltk.stem import PorterStemmer
from nltk.stem import SnowballStemmer
from nltk.stem import LancasterStemmer

porter = PorterStemmer()
snowball = SnowballStemmer("english")   
lancaster = LancasterStemmer()

words = ["running", "ran", "runs", "easily", "fairly", "happily", "happiness"]

# stopwords = set(stopwords.words('english')) 

for word in words:
    print(f"Original: {word}")
    print(f"Porter Stemmer: {porter.stem(word)}")
    print(f"Snowball Stemmer: {snowball.stem(word)}")
    print(f"Lancaster Stemmer: {lancaster.stem(word)}")
    print()
```

```python
from nltk.stem import WordNetLemmatizer

# nltk.download('wordnet')
# nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

words = ["running", "ran", "runs", "easily", "fairly", "happily", "happiness", "tracking", "delayed"]

lemmatized = [
    lemmatizer.lemmatize(word)
    for word in words
] 
print(lemmatized)
    
```
