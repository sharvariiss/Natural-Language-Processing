## Text Preprocessing Pipeline ## 

```python
import nltk 
import spacy
import re
```

```python
nltk.download("punkt")
nltk.download("stopwords")
```

```python
text = """
Natural Language Processing, also known as NLP, is one of the most important areas of Artificial Intelligence.
Today, companies use NLP in chatbots, search engines, recommendation systems, social media analytics,
customer feedback analysis, automatic translation, and voice assistants.
 
In 2026, many organizations are collecting huge amounts of text data from websites, emails, reviews,
online forms, tweets, product comments, and support tickets. However, this text data is often messy.
For example, a customer may write: "I LOVE this product!!! Visit https://example.com/product123 #Amazing #AI".
Another user may write: "The service was not good... I waited 45 minutes and no one replied."
 
Before applying machine learning or deep learning models, we need to clean and prepare the text.
This step is called text preprocessing. It may include converting text to lowercase, removing URLs,
removing punctuation, removing numbers, tokenizing sentences into words, removing stop words,
stemming, lemmatization, and sometimes removing emojis or special characters.
 
Good preprocessing can improve the quality of NLP models. However, preprocessing should be done carefully.
For example, removing the word "not" from the sentence "I am not happy" can change the meaning completely.
Similarly, removing hashtags may not always be a good idea in social media analytics because hashtags
can contain useful information about topics, opinions, or trends.
 
Your task is to build a complete NLP preprocessing pipeline for this text.
You should clean the text step by step and observe how the output changes after each step.
"""
```

```python
## Cleaning the Text ##

cleaned_text = re.sub(r'[^\w\s]', '', text) #Punctuation removal
cleaned_text = re.sub(r'http\S+|www\S+', '', cleaned_text) #URL removal
cleaned_text = re.sub(r'#[\w]+', '', cleaned_text) #Hashtag removal
cleaned_text = re.sub(r"[^\x00-\x7F]+", "", cleaned_text) # remove emojis/non-ascii

print("CLEANED TEXT:", cleaned_text)
```

```python
## Tokenization ##

from nltk.tokenize import word_tokenize

tokenize = word_tokenize(cleaned_text)
print("TOKEN:", {token for token in tokenize})
```

```python
## Removing Stopwords, punctuations and numbers ##

from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

fiterered_tokens = [
    token for token in tokenize if token.isalpha() and token not in stop_words
]
print("FILTERED TOKENS:", fiterered_tokens) 
```

```python
## Lemmatization using NLTK ##

from nltk.stem import WordNetLemmatizer

filtered_tokens = ['Natural', 'Language', 'Processing', 'also', 'known', 'NLP', 'one', 'important', 'areas', 'Artificial', 'Intelligence', 'Today', 'companies', 'use', 'NLP', 'chatbots', 'search', 'engines', 'recommendation', 'systems', 'social', 'media', 'analytics', 'customer', 'feedback', 'analysis', 'automatic', 'translation', 'voice', 'assistants', 'In', 'many', 'organizations', 'collecting', 'huge', 'amounts', 'text', 'data', 'websites', 'emails', 'reviews', 'online', 'forms', 'tweets', 'product', 'comments', 'support', 'tickets', 'However', 'text', 'data', 'often', 'messy', 'For', 'example', 'customer', 'may', 'write', 'I', 'LOVE', 'product', 'Visit', 'Amazing', 'AI', 'Another', 'user', 'may', 'write', 'The', 'service', 'good', 'I', 'waited', 'minutes', 'one', 'replied', 'Before', 'applying', 'machine', 'learning', 'deep', 'learning', 'models', 'need', 'clean', 'prepare', 'text', 'This', 'step', 'called', 'text', 'preprocessing', 'It', 'may', 'include', 'converting', 'text', 'lowercase', 'removing', 'URLs', 'removing', 'punctuation', 'removing', 'numbers', 'tokenizing', 'sentences', 'words', 'removing', 'stop', 'words', 'stemming', 'lemmatization', 'sometimes', 'removing', 'emojis', 'special', 'characters', 'Good', 'preprocessing', 'improve', 'quality', 'NLP', 'models', 'However', 'preprocessing', 'done', 'carefully', 'For', 'example', 'removing', 'word', 'sentence', 'I', 'happy', 'change', 'meaning', 'completely', 'Similarly', 'removing', 'hashtags', 'may', 'always', 'good', 'idea', 'social', 'media', 'analytics', 'hashtags', 'contain', 'useful', 'information', 'topics', 'opinions', 'trends', 'Your', 'task', 'build', 'complete', 'NLP', 'preprocessing', 'pipeline', 'text', 'You', 'clean', 'text', 'step', 'step', 'observe', 'output', 'changes', 'step']

lemmatize = WordNetLemmatizer()

lemmatized_tokens = [lemmatize.lemmatize(token) for token in filtered_tokens]

print("LEMMATIZED TOKENS:", lemmatized_tokens)
```

```python
## Lemmatization using spaCy ##

# spacy.cli.download("en_core_web_sm")

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

filtered_text = " ".join(filtered_tokens)

doc = nlp(filtered_text)

lemmatized_tokens = [
    token.lemma_
    for token in doc
]

print("LEMMATIZED TOKENS:")
print(lemmatized_tokens)
```

```python
## Final Output ## 

final_text = " ".join(lemmatized_tokens)

print("\nFINAL PREPROCESSED TEXT:")
print(final_text)
```
