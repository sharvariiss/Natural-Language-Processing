```python
import spacy
from spacy import displacy
from IPython.display import HTML, display

# spacy.cli.download("en_core_web_sm")
nlp = spacy.load("en_core_web_sm")

text = "I saw the man with a telescope."

doc = nlp(text)

for token in doc:
    print(f"{token.text:<12} {token.dep_:<12} {token.head.text}")

html = displacy.render(doc, style="dep", jupyter=False)

display(HTML(html))
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = """
Microsoft announced a new AI research initiative
in Berlin on 15 June 2026.

The project involves collaboration with Google,
SAP and SRH University Heidelberg.

Satya Nadella will present the initiative
at a conference in Germany.
"""

doc = nlp(text)

for ent in doc.ents:
    print(f"Entity Text   : {ent.text}")
    print(f"Entity Label  : {ent.label_}")
    print(f"Explanation   : {spacy.explain(ent.label_)}")
    print("-" * 40)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "AI is transforming healthcare",
    "Machine learning improves healthcare",
    "Football is a popular sport"
]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())

similarity_matrix = cosine_similarity(tfidf_matrix)

print("\nCosine Similarity Matrix:")
print(similarity_matrix)

print("Doc1 vs Doc2:", round(similarity_matrix[0][1], 3))
print("Doc1 vs Doc3:", round(similarity_matrix[0][2], 3))
print("Doc2 vs Doc3:", round(similarity_matrix[1][2], 3))
```

# 1. Which pair has the highest similarity?
    Highest similarity:
        Doc1 vs Doc2 = 0.172
        Doc1 vs Doc3 = 0.172

# 2. Is this result reasonable?
    Partially yes.
    - Doc1 and Doc2 are related because both discuss healthcare.
    - Doc1 and Doc3 share only the word “is”.
    TF-IDF gives similarity based on shared words, not true meaning.

# 3. Why does TF-IDF treat some words as more important?
    TF-IDF gives higher importance to:
    - words that appear frequently in one document
    - but rarely across all documents
    Common words get lower importance.

# 4. Does cosine understand meaning?
    No.
    Cosine similarity only compares vector patterns and word frequencies.
    It does not truly understand semantics or context.

# 5. Would it know that "AI" and "machine learning" are related?
    No.
    TF-IDF treats them as completely different words unless they literally overlap.
    To understand semantic relationships, we need:
    - Word embeddings
    - Word2Vec
    - GloVe
    - BERT
    - Sentence Transformers
    These models capture meaning, unlike TF-IDF.
